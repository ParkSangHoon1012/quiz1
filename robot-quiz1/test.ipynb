```python
def delivery_fee(total, distance):
    def main():
        total = int(input("주문금액[원] : "))
        distance = int(input("거리[km] : "))
        if 30000 <= total:
            return 0
            if 2 > distance:
                return 2000
            elif 2 <= distance < 5:
                return 3000
            elif 5 <= distance:
                return 5000
        else:
            return

bedal = input(delivery_fee)
print(f"배달 요금 {bedal}원")
```
